```python
import requests
from bs4 import BeautifulSoup
import pandas as pd

url = "https://search.naver.com/search.naver"

params = {
    "where": "news",
    "query": "인공지능"
}

headers = {
    "User-Agent": "Mozilla/5.0"
}

response = requests.get(
    url,
    params=params,
    headers=headers,
    timeout=10
)

response.raise_for_status()

soup = BeautifulSoup(response.text, "html.parser")

# 뉴스 목록
news_list = soup.select(".fds-news-item-list-tab > div")

data = []

for news in news_list:

    title = news.select_one(
        'a[data-heatmap-target=".tit"]'
    )

    description = news.select_one(
        'a[data-heatmap-target=".body"]'
    )

    press = news.select_one(
        ".sds-comps-profile-info-title-text"
    )

    if title is None:
        continue

    data.append({
        "제목": title.get_text(" ", strip=True).replace("새 창 열림", "").strip(),
        "언론사": press.get_text(" ", strip=True).replace("새 창 열림", "").strip() if press else "",
        "요약": description.get_text(" ", strip=True).replace("새 창 열림", "").strip() if description else "",
        "URL": title.get("href")
    })

df = pd.DataFrame(data)

display(df)

df.to_excel("네이버뉴스.xlsx", index=False)
```

```python
from selenium import webdriver
from bs4 import BeautifulSoup
import pandas as pd
import time

driver = webdriver.Chrome()

url = "https://search.naver.com/search.naver?where=news&query=인공지능"

driver.get(url)

time.sleep(2)

# 스크롤 5번
for i in range(5):
    driver.execute_script(
        "window.scrollTo(0, document.body.scrollHeight);"
    )
    time.sleep(2)

# 현재 브라우저의 HTML 가져오기
html = driver.page_source

soup = BeautifulSoup(html, "html.parser")

news_list = soup.select(".fds-news-item-list-tab > div")

data = []

for news in news_list:

    title = news.select_one(
        'a[data-heatmap-target=".tit"]'
    )

    description = news.select_one(
        'a[data-heatmap-target=".body"]'
    )

    press = news.select_one(
        ".sds-comps-profile-info-title-text"
    )

    if title is None:
        continue

    data.append({
        "제목": title.get_text(" ", strip=True).replace("새 창 열림", "").strip(),
        "언론사": press.get_text(" ", strip=True).replace("새 창 열림", "").strip() if press else "",
        "요약": description.get_text(" ", strip=True).replace("새 창 열림", "").strip() if description else "",
        "URL": title.get("href")
    })

df = pd.DataFrame(data)

df = df.drop_duplicates(subset=["URL"])

display(df)

df.to_excel("네이버뉴스.xlsx", index=False)

driver.quit()
```
