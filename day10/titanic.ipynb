```python
import pandas as pd
df = pd.read_csv("train.csv")
# display(df.head())
# display(df.tail())
# print(df.shape)
# print(df.columns)
df.info()
```

```python
result = df.isnull()
# display(result.head())
result = df.isnull().sum()
# print(result)
condition = df["Age"].isnull()
result = df[condition]
display(result[["Name","Age","Sex"]])
```

```python
# 결측데이터가 행에 하나라도 있으면 삭제하기 5% 미만일 경우 삭제하기도 한다.
result = df.dropna()
# display(result.shape)
result = df.dropna(subset=["Age"])
# display(result.shape)
result = df.copy()
# age에 값이 없으면 0으로 채워라
result["Age"] = result["Age"].fillna(0)
display(result[["Name","Age"]])
```

```python
# 결측을 평균으로 채우기 이건 데이터가 극단적인 경우 좋지 않은 방법이다.
result = df.copy()
age = result["Age"]
avg = age.mean()
result["Age"] = age.fillna(avg)
display(result[["Name","Age"]])
```

```python
# 결측을 중앙값으로 채우기 
result = df.copy()
age = result["Age"]
median = age.median()
result["Age"] = age.fillna(median)
display(result[["Name","Age"]])
```

```python
# 문자데이터 결측을 최빈값(최고 빈도)으로 채우기 즉 제일 많이 나타나는 빈도수로 채우기
result = df.copy()
mode = result["Embarked"].mode()[0]
result["Embarked"] = result["Embarked"].fillna(mode)
display(result.isnull().sum())
# display(mode)


scores = pd.Series([30,30,30,30,30,30,30,10,10,10,20,20,20,20])
result = scores.mode()
print(result)

result02 = scores.value_counts()
result02 = result02.idxmin()
print(result02)

```
