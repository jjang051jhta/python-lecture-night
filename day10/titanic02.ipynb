```python
import pandas as pd
df = pd.read_excel("titanic_cleaned.xlsx")
# display(df.head())
print(df.shape)
result = df["Survived"].value_counts()
print("사망자수: ",result[0])
print("생존자수: ",result[1])
result = df["Survived"].mean()*100
print(f"생존율: {result:.2f}%")

# 남녀 생존율
result = df.groupby("Sex")["Survived"].mean()*100
display(result.round(2))

# 객실 등급 생존율
result = df.groupby("Pclass")["Survived"].mean()*100
display(result.round(2))

# 남성 여성
result = df.groupby("Sex")["Age"]
result = result.agg(["count","mean","min","max"])
display(result)



```
