```python
# pip install matplotlib  설치

import pandas as pd
import matplotlib.pyplot as plt
df = pd.read_excel("titanic_cleaned.xlsx")
# display(df.head())
result = df["Sex"].value_counts()
display(result)
plt.rcParams["font.family"]="Malgun Gothic"
result.plot(kind="bar",figsize=(6,4),color=["#a213ce","#9b9807"])
plt.title("타이타닉 승객 남녀 비율",fontsize=18,pad=20)
plt.xlabel("성별")
plt.ylabel("승객수")
# plt.tight_layout()
# plt.subplots_adjust(top=0.5)
# 파일로 차트 저장

plt.savefig("타이타닉 남녀생존율.png") 
plt.show()
```

```python
# Pclass 객실 등급별 승객수
result = df["Pclass"].value_counts().sort_index()
result.index=["1등실","2등실","3등실"]
display(result)
result.plot(kind="bar")
plt.title("객실 등급별 승객수")
plt.xlabel("객실등급")
plt.ylabel("승객수")
plt.xticks(rotation=0, ha="right")

plt.show()
# Embarked 승객수 S,Q,C 그래프 그리기
```

```python
result = df["Embarked"].value_counts().sort_index()
# result.index=["1등실","2등실","3등실"]
display(result)
# result.index=["Queenstown","Southhampton","Cherboug"]
# result = result.reindex(["S","Q","C"])
# result.rename({"S":"Southhampton","C":"Cherboug","Q":"Queenstown"})
axis = result.plot(kind="bar")
axis.set_axisbelow(True)
axis.bar_label(axis.containers[0],padding=1,fontsize=9)
plt.title("탑승항구별 승객구")
plt.xlabel("탑승항구")
plt.ylabel("승객수")
plt.xticks(rotation=0)

plt.grid(axis="y",linestyle="--",alpha=0.5)
plt.show()
# Embarked 승객수 S,Q,C 그래프 그리기
```

```python
result = df["Embarked"].value_counts().sort_values(ascending=False)
# result.index=["1등실","2등실","3등실"]
display(result)
# result.index=["Queenstown","Southhampton","Cherboug"]
result = result.reindex(["S","Q","C"])
result.rename({"S":"Southhampton","C":"Cherboug","Q":"Queenstown"})
axis = result.plot(kind="bar")
axis.set_axisbelow(True)
axis.bar_label(axis.containers[0],padding=1,fontsize=9)
plt.title("탑승항구별 승객구")
plt.xlabel("탑승항구")
plt.ylabel("승객수")
plt.xticks(rotation=0)

plt.grid(axis="y",linestyle="--",alpha=0.5)
plt.show()
# Embarked 승객수 S,Q,C 그래프 그리기
```

```python
result = df["Sex"].value_counts()
result.index=["남자","여자"]
result.plot(
    kind="bar",
    figsize=(8, 5),
    color=["skyblue", "salmon"]
)

plt.title("타이타닉 성별 승객 수", fontsize=18, pad=20)

plt.xlabel("성별")
plt.ylabel("승객 수")

# 아래쪽 X축
plt.xticks(rotation=45, ha="right", fontsize=12)

# 왼쪽 Y축
plt.yticks(
    [0, 100, 200, 300, 400, 500, 600],
    fontsize=12,
    rotation=45
)

plt.tight_layout()
plt.show()
```
