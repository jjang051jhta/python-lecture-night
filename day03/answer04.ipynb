```python
def getMax(a,b):
    if a>b:
        return a
    return b
print(getMax(100,20))
# 함수는 return을 만나면 더 이상 실행하지  않고 사라진다.
```

```python
def check_num(num):
    if num%2==0:
        return "짝수"
    return "홀수"
print(check_num(42))
print(check_num(31))
```

```python
# 문제 7. 1부터 입력받은 숫자까지의 합
def sum_number(num):
    total=0
    for i in range(1,num+1):
        total = total+i
    return total
print(sum_number(1212))
```

```python
def gugudan(num):
    for i in range(1,10):
        print(f"{num} x {i} = {num*i}")
gugudan(19)
```

```python
# 1부터 입력받은 숫자까지 짝수의 합
def sum_even(num):
    total = 0
    for i in range(1,num+1):
        if i%2==0:
            total=total+i

    return total
print(sum_even(4324))
```

```python
def get_grade(num):
    if num>=90:
        return "A"
    elif num>=80:
        return "B"
    elif num >= 70:
        return "C"
    elif num >= 60:
        return "D"
    return "F"
print(get_grade(88))
        
```
