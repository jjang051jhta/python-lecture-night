```python
df = pd.read_csv("아파트(매매)_실거래가_20261014203320.csv",
                 encoding="cp949",
                 skiprows=15)
# print(df.head(10))
# print("==="*30)
# print(df.tail(10))
# print("==="*30)
print(df.columns)
# print("==="*30,"shape")
# print(df.shape)
# print("/////"*30,"info")
# print(df.info())
df["거래금액(만원)"]=(df["거래금액(만원)"].str.replace(",","",regex=False)).astype(int)
df[["단지명","전용면적(㎡)","거래금액(만원)","층"]]
result = df.sort_values(by="거래금액(만원)",ascending=False)
# print(result)

top10 = result[["단지명","전용면적(㎡)","거래금액(만원)"]].head(10)
print(top10)
print("===="*30)
avg = df.groupby("단지명")["거래금액(만원)"].mean()
print(avg)
avg02 = avg.sort_values(ascending=False)
avg02.to_excel("강남아파트 단지별 평균금액_내림차순.xlsx")
```
