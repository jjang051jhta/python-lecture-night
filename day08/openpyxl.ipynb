```python
import csv
from openpyxl import Workbook
wb = Workbook()
ws = wb.active
ws.title = "판매데이터"
with open("판매데이터_실습용.csv","r",
          encoding = "utf-8") as file:
    reader = csv.reader(file)
    for row in reader:
        ws.append(row)
wb.save("step01.xlsx")        
print("엑셀 파일 생성 완료")
```

```python
import csv
from openpyxl import Workbook
wb = Workbook()
ws = wb.active
ws.title = "판매금액"
with open("판매데이터_실습용.csv","r",
          encoding = "utf-8") as file:
    reader = csv.DictReader(file)
    ws.append(["상품명","수량","단가","판매금액"])
    for row in reader:
        product = row["상품명"]
        quantity = int(row["수량"])
        price = int(row["단가"])
        total = quantity*price
        ws.append([product,quantity,price,total])
wb.save("step02.xlsx")        
print("엑셀 파일 생성 완료")
```
