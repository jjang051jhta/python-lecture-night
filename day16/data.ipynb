```python
import folium
lat = 37.566535
longi = 126.977969
m = folium.Map(location=[lat,longi],zoom_start=13)
folium.Marker(location=[lat,longi], 
              popup=folium.Popup('<div style="white-space:nowrap">서울시청</div>'),
              tooltip="여기를 클릭하세요").add_to(m)
m.save("seoul.html")

```

```python
# 위도: 37.579982° 
# 경도: 126.977153° 

 
# 위도: 37.551300° 
# 경도: 126.988230° 

places = [
            {"name":"서울시청","lat":37.566535, "longi":126.977969},
            {"name":"경복궁","lat":37.579982, "longi":126.977153},
            {"name":"남산타워","lat":37.551300, "longi":126.988230}
          ]
m = folium.Map(location=[places[0]["lat"],places[0]["longi"]],zoom_start=13)
for place in places:
    folium.Marker(location=[place["lat"],place["longi"]], 
              popup=place["name"],
              tooltip=place["name"]).add_to(m)
m.save("seoul02.html")
```
