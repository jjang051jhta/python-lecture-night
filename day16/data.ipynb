```python
# 위도: 37.579982° 
# 경도: 126.977153° 

 
# 위도: 37.551300° 
# 경도: 126.988230° 
import folium
from folium.plugins import MarkerCluster, HeatMap



places = [
            {"name":"서울시청","lat":37.566535, "longi":126.977969},
            {"name":"경복궁","lat":37.579982, "longi":126.977153},
            {"name":"남산타워","lat":37.551300, "longi":126.988230},
            {"name":"덕수궁","lat":37.566111, "longi":126.975000},
            {"name":"명동성당","lat":37.5634, "longi":126.9873},

          ]
m = folium.Map(location=[places[0]["lat"],places[0]["longi"]],zoom_start=13)
cluster  = MarkerCluster().add_to(m)
for place in places:
    folium.Marker(location=[place["lat"],place["longi"]], 
              popup=folium.Popup(f'<div style="white-space:nowrap">{place["name"]}</div>'),
              tooltip=place["name"]).add_to(cluster)
heat_data=[]
for place in places:
    heat_data.append([place["lat"], place["longi"]])
HeatMap(heat_data, radius=30, blur=20, min_opacity=0.3).add_to(m)

m.save("seoul04.html")
```

```python
import requests
import json
import pandas as pd

all_parking=[]
page=1

url = "https://apis.data.go.kr/6260000/BusanPblcPrkngInfoService/getPblcPrkngInfo"
while True:
    my_params = {
        "serviceKey":"디코딩된 키",
        "numOfRows":100,
        "pageNo":page,
        "resultType":"json"
    }

    response = requests.get(url,params=my_params, timeout=20)
    response.raise_for_status()
    data = response.json()
    body = data["response"]["body"]
    items = body["items"]["item"]
    total_count = int(body["totalCount"])
    all_parking.extend(items)
    if len(all_parking) >= total_count:
        break
    page += 1

print("전체 데이터 수집 완료")
print("전체 데이터 수집 갯수",len(all_parking))

df = pd.DataFrame(all_parking)
df.to_excel("busan_parking.xlsx", index=False)
```
