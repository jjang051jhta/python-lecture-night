```python
import os
sites = {
    "naver": "https://www.naver.com",
    "google": "https://www.google.com",
    "daum": "https://www.daum.net",
}
os.makedirs("qrcodes",exist_ok=True)
for name,url in sites.items():
    qrImg = qrcode.make(url)
    qrImg.save(f"qrcodes/{name}.png")
    print(f"{name}.png 생성완료")
```

```python
import os
from PIL import Image
qr =qrcode.QRCode(version=1,
    error_correction=qrcode.constants.ERROR_CORRECT_H,
    box_size=10,
    border=2
)
qr.add_data("http://www.naver.com")
qr.make(fit=True)
qrImg = qr.make_image(fill_color="black",back_color="white").convert("RGBA")
logo = Image.open("apple.png").convert("RGBA")
qrWidth,qrHeight = qrImg.size
logoSize=qrWidth//5
logo.thumbnail((logoSize,logoSize))
x=(qrWidth-logo.width)//2
y=(qrHeight-logo.height)//2
qrImg.paste(logo,(x,y),logo)
qrImg.save("my_qrcode.png")
print("로고 들어간 qrcode완성")

```
