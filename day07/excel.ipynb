```python
from openpyxl import Workbook

wb = Workbook()
ws = wb.active
ws.title = "학생명단"
total =0
count = 0
with open("students.txt","r",encoding="utf-8") as file:
    header = file.readline().strip().split(",")
    ws.append(header)
    for line in file:
        name,age,region = line.strip().split(",")
        age = int(age)
        ws.append([name,age,region])
        total += age
        count += 1
avg =  total/count        
ws.append(["평균나이",avg,""])
        # ws.append(data)
wb.save("students03.xlsx")        
print("학생명단 저장완료")
```
