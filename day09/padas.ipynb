```python
# pip install pandas
import pandas as pd
# encoding="utf-8"로 읽으면 되는데 가끔 한글의 경우
# 오류가 날때가 있다 이때 cp949로 읽으면 된다.
df = pd.read_csv("아파트(매매)_실거래가_20261019190916_일산동구.csv",
            encoding="cp949",
            skiprows=15,
            thousands=","
            )
print(df)
```
