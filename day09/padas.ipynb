```python
year  = df["건축년도"]
condition = year < 1995
result = df[condition]
result = result[["단지명","거래금액(만원)","건축년도"]]
result = result.sort_values("건축년도",ascending=True)
result.to_excel("건축년도.xlsx",index=False)
display(result)
```

```python
price  = df["거래금액(만원)"]
condition = price < 50000
print(df[condition])
```

```python
# 거래가 많이 발생한 아파트
result = df["단지명"].value_counts()
display(result.head(10))
```

```python
# 전용면적이 84이상이면서 거래금액이 8억 이상인 아파트
area = df["전용면적(㎡)"] >= 84
price = df["거래금액(만원)"] >= 80000
condition = area & price
result = df[condition]
result = result[["단지명","전용면적(㎡)","거래금액(만원)"]]
display(result)
```

```python
# 아파트별 최고가 최저가 평균가
# result = df.groupby("단지명")["거래금액(만원)"].agg(["max","min","mean","count"])
# result = result.sort_values("mean",ascending=False)
result = df.groupby("단지명")["거래금액(만원)"].agg(최고가 = "max",최저가="min",평균가="mean",거래건수="count")
result = result.sort_values("평균가",ascending=False)
result.to_excel("아파트.xlsx",sheet_name="아파트거래")
display(result)
```

```python
# 시트별 개별 저장
price = df["거래금액(만원)"]
result01 =  df[price>=100000]

year = df["건축년도"]
result02 =  df[year < 1995]


result03 = df.groupby("단지명")["거래금액(만원)"].agg(최고가 = "max",최저가="min",평균가="mean",거래건수="count")
result03 = result03.sort_values("평균가",ascending=False)

with pd.ExcelWriter("아파트분석.xlsx",engine="openpyxl") as writer:
    df.to_excel(writer,sheet_name="전체거래", index=False)
    result01.to_excel(writer,sheet_name="10억이상", index=False)
    result02.to_excel(writer,sheet_name="1995년 이전", index=False)
    result03.to_excel(writer,sheet_name="아파트 평균 최고가 최저가", index=False)
        

```
