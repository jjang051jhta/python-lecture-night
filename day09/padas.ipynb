```python
# pip install pandas
import pandas as pd
# encoding="utf-8"로 읽으면 되는데 가끔 한글의 경우
# 오류가 날때가 있다 이때 cp949로 읽으면 된다.
df = pd.read_csv("아파트(매매)_실거래가_20261019190916_일산동구.csv",
            encoding="cp949",
            skiprows=15,
            thousands=","
            )
print(df)
```

```python
# result = df[["단지명","전용면적(㎡)","거래금액(만원)"]]
columns = ["단지명","전용면적(㎡)","거래금액(만원)"]
result = df[columns]
result = result.sort_values("거래금액(만원)",ascending=False)
print(result.head(10))
```

```python
price = df["거래금액(만원)"]
# print(price)
condition = price >= 100000
result = df[condition]
# result = result[["단지명","전용면적(㎡)","거래금액(만원)"]]
# print(result.tail(10))
# 표 형태로 보기
display(result.tail(10))
```

```python
result = df[df["단지명"].str.contains("킨텍스")]
# result = df[df["단지명"]=="킨텍스"]
# display(result[["단지명","거래금액(만원)"]])
result = result.sort_values(by="단지명")
result.to_excel("킨텍스_아파트.xlsx")
```

```python
result = df[df["전용면적(㎡)"].between(80,100)]
display(result[["단지명","전용면적(㎡)","거래금액(만원)"]])
```

```python
price = df["거래금액(만원)"]
print("평균 거래금액",price.mean())
print("최고가 거래금액",price.max())
print("최저 거래금액",price.min())
```
