```python
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.common.keys import Keys
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC
import pandas as pd


driver = webdriver.Chrome()
driver.get("https://search.naver.com/search.naver")

search = driver.find_element(By.NAME,"query")
search.send_keys("트럼프 중간선거")
search.send_keys(Keys.ENTER)
wait = WebDriverWait(driver,10)
more = wait.until(EC.element_to_be_clickable((By.CSS_SELECTOR,'a[data-heatmap-target=".more"]')))
more.click()

wait.until(EC.presence_of_element_located((By.CSS_SELECTOR,".fds-news-item-list-tab > div")))
for i in range(10):
    driver.execute_script("window.scrollTo(0,document.body.scrollHeight);")
    time.sleep(2)

news_list = driver.find_elements(By.CSS_SELECTOR,".fds-news-item-list-tab > div")
print(len(news_list))

data = []
for news in news_list:
    titles = news.find_elements(By.CSS_SELECTOR,'a[data-heatmap-target=".tit"]')
    desces = news.find_elements(By.CSS_SELECTOR,'a[data-heatmap-target=".body"]')
    # presses = news.find_elements(By.CSS_SELECTOR,'a[data-heatmap-target=".body"]')
    if not titles:
        continue
    title = titles[0].text
    desc = desces[0].text
    data.append({"제목":title,"내용":desc})
df = pd.DataFrame(data)
df.to_excel("네이버뉴스.xlsx",index=False)
driver.quit()
        
        
```
