````python
# 구구단 출력
# 3x1=3*1
# 3x2=3*2
# 3x3=3*3
num = int(input("구구단 출력 원하는 숫자를 입력하세요"))
for i in range(1,10):
    print(num," x ",i,"=",num*i)
````

````python
num = int(input("정수를 입력하세요: "))

sum = 0

for i in range(1, num + 1):
    sum = sum + i
    if i < num:
        print(i,end="+")
    else:
        print(i, end="")
print(" =", sum)
````
