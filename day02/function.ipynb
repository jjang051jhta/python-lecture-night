```python
def hello():
    print("hello world")
    print("hello python")
    print("hello 장성호")

for i in range(0,1):
    hello()

# 함수를 매개해주는 변수를 매개변수 (parameter)
def plus(x,y):
    print(x," + ",y ,"=",x+y)  

def minus(x,y):
    print(x," - ",y ,"=",x-y)  
    return x-y

def multiple(x,y):
    return x*y


myPrint=print("aaaa")
print(myPrint)

myPlus = plus(10,3)  # 함수의 인자 argument   None
print(myPlus)
plus(10,4)
plus(10,5)
result = minus(10,5)
print(result)
print(multiple(3,4)) #함수를 값으로 바꿔서 출력하고 싶다.


# width*height/2
def triangle(w,h):
    return w*h/2
    
print(triangle(10,4))

# 함수의 4가지 형태
# 1. 매개변수 x / return x
# 1. 매개변수 o / return x
# 1. 매개변수 x / return 0
# 1. 매개변수 0 / return 0
def gugudan(num):
    for i in range(1,10):
        print(num," x ",i,"=",num*i)
    


gugudan(2)
gugudan(3)
gugudan(4)



```
