```python
import pandas as pd
import matplotlib.pyplot as plt
df = pd.read_csv("apt.csv",encoding="cp949",thousands=",",skiprows=15)
# display(df.head())
# df["거래금액(만원)"]=(df["거래금액(만원)"]
#                     .astype(str)
#                     .str
#                     .replace(",","",regex=False))
# print(df["거래금액(만원)"].head())
plt.rcParams["font.family"]="Malgun Gothic"
df["거래금액(만원)"] = pd.to_numeric(df["거래금액(만원)"], errors="coerce")
# NaN
result = df.groupby("단지명")["거래금액(만원)"].mean()
result = result.sort_values(ascending=False).head(10)
# display(result.head())
axis = result.plot(kind="bar", figsize=(1200/72, 600/72))
axis.figure.set_dpi(72)
axis.grid(axis="y")
axis.set_axisbelow(True)
plt.title("2026년 일산동구 아파트 단지별 평균 거래 금액")
# plt.show()



```

```python
result = df.groupby("계약년월")["거래금액(만원)"].mean()
result = result.sort_index()
result.index = result.index.astype(str)
result.plot(kind="line", marker="s", linewidth=1)
plt.title("2026년 월별 아파트 평균 거래금액")
plt.grid(alpha=0.3, axis="y")

```

```python
result = df["거래금액(만원)"]
result.plot(kind="hist", bins=50)
plt.title("아파트 거래금액 분포도")
plt.grid(alpha=0.3, axis="y")
```
