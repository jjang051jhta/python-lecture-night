```python
import random
num = random.randint(1,46)
print(num)
for i in range(5):
    lotto = []
    while len(lotto) < 6:
        num = random.randint(1,45)
        if num not in lotto:
            lotto.append(num)
# print("로또번호 : ",lotto)            
    lotto.sort()        
    print(i+1,"번째 게임 : ",lotto)    
print("==="*30)
for i in range(5):
    lotto=random.sample(range(1,46),6)
    lotto.sort()
    print(i+1,"번째 게임 : ",lotto)
print("==="*30)




```

```python
computer = random.choice(["가위","바위","보"])    
human = input("가위, 바위, 보 중에 하나를 입력하세요")
print("나 : ",human)
print("컴퓨터 : ",computer)
if human == computer:
    print("무승부")
elif computer == "가위" and human == "바위":
    print("이겼습니다.")   
elif computer == "바위" and human == "보":
    print("이겼습니다.")   
elif computer == "보" and human == "가위":
    print("이겼습니다.")   
else:
    print("졌습니다.")       


# print(computer)
```

```python
answer = random.randint(1,100)
print(answer)
while True:
    num = int(input("1~100사이의 숫자를 입력하세요"))
    if num > answer :
        print("DOWN")
    elif num < answer:
        print("UP")
    else:
        print("딩동댕")
        break
```
